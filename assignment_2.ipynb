```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
```

```python
def generate_product_description(product_name):
    prompt = (
        f"Product Name: {product_name}\n"
        f"Writing a compelling marketing description highlighting features and benefits:\n"
    )

    output = generator(
        prompt,
        max_length=120,
        num_return_sequences=1,
        do_sample=True,
        temperature=0.9,
        top_p=0.95
    )

    return output[0]["generated_text"]
```

```python
product_name = "playstation 5"
description = generate_product_description(product_name)

print(description)
```
